## Python


```python


from typing import List

from ast import List


class LeetCode1:

    def twoSum(self, nums: List[int], target: int) -> List[int]:
        # Defining the function to find the two indices of the numbers that add up to the target value
        acumulator = {}
        # Gatheing the numbers and their indices in a dictionary for quick lookup

        for index, num in enumerate(nums):
            # Enumerate allows us to loop through the list while keeping track of the index and the number at that index
            # index is the position of the number in the list, and num is the actual number at that position
            complement = target - num

            if complement in acumulator:
                # If the complement of the current number is in the dictionary, we have found our pair
                result = [acumulator[complement], index]
                return result
            
                # We return the indices of the two numbers that add up to the target value
                # acumulator[complement] gives us the index of the complement number, and index is the current index of the number we are checking
            acumulator[num] = index
```

```python
%%writefile program.c
#include <stdio.h>
#include <stdlib.h>

int* twoSum(int* nums, int numsSize, int target, int* returnSize) 
    # Defining a function that takes an array of integers (nums), the size of the array (numsSize), a target integer (target), and a pointer to an integer (returnSize) as input. The function returns a pointer to an array of two integers.
    # int* twoSum defines a function that returns a pointer to an integer array. The function takes four parameters: an integer array (nums), the size of the array (numsSize), a target integer (target), and a pointer to an integer (returnSize).
    # int numsSize is the size of the input array, which is used to control the loop that searches for the two numbers.
    # int* returnSize is a pointer to an integer that will be used to store the size of the result array. It is set to 2 if a valid pair is found, or 0 if no pair is found.
{
    int* result = malloc(2 * sizeof(int));
    # Allocating memory for an array of two integers to store the indices of the two numbers that add up to the target value. The malloc function is used to allocate memory dynamically, and sizeof(int) gives the size of an integer in bytes.
    for (int i = 0; i < numsSize; i++) 
    # Looping through the array of integers using a for loop. The loop variable i starts at 0 and increments by 1 until it reaches numsSize - 1. This loop is used to select the first number in the pair.
    
    {
        for (int j = i + 1; j < numsSize; j++) 
        # Nested loop to select the second number in the pair. The loop variable j starts at i + 1 to ensure that we do not use the same number twice and increments by 1 until it reaches numsSize - 1.
        {
            if (nums[i] + nums[j] == target)
            # Checking if the sum of the two selected numbers (nums[i] and nums[j]) is equal to the target value. If it is, we have found the pair of numbers that add up to the target.
            {
                result[0] = i;
                result[1] = j;
                # Storing the indices of the two numbers in the result array. result[0] is set to i (the index of the first number), and result[1] is set to j (the index of the second number).
                *returnSize = 2;
                return result;
                # Setting the value pointed to by returnSize to 2, indicating that the result array contains two indices. The function then returns the result array containing the indices of the two numbers that add up to the target value.
            }
        }
    }

    *returnSize = 0;
    free(result);
    return NULL;
    # If no pair of numbers is found that adds up to the target value, the function sets the value pointed to by returnSize to 0, frees the memory allocated for the result array, and returns NULL to indicate that no valid pair was found.
}

int main(void) {
    int nums[] = {2, 7, 11, 15};
    # Defining an array of integers (nums) with the values 2, 7, 11, and 15. This array will be used as input to the twoSum function.
    int returnSize;
    # Declaring an integer variable (returnSize) to store the size of the result array returned by the twoSum function. This variable will be used to check if a valid pair of indices was found.
    int* result = twoSum(nums, 4, 9, &returnSize);
    # Calling the twoSum function with the nums array, its size (4), the target value (9), and the address of returnSize. The result of the function call is stored in the result pointer, which will point to an array containing the indices of the two numbers that add up to the target value if a valid pair is found.
    if (result != NULL) 
    {
        printf("[%d, %d]\n", result[0], result[1]);
        free(result);
        # If a valid pair of indices is found (result is not NULL), the program prints the indices in the format [index1, index2]. It then frees the memory allocated for the result array to prevent memory leaks.
    }

    return 0;
    # Returning 0 from the main function to indicate that the program has executed successfully. This is a standard practice in C programming to signal that the program has completed without errors.
}
```
